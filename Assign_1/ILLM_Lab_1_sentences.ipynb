```python
import re
import math
import random
import numpy as np
import pandas as pd

from collections import Counter, defaultdict
from datasets import load_dataset
```

```python
from datasets import load_dataset

dataset = load_dataset(
    "ai4bharat/IndicCorpV2",
    streaming=True
)
```

```python
print(dataset)
```

```python
languages = [
    "asm_Beng",
    "ben_Beng",
    "brx_Deva",
    "doi_Deva",
    "gom_Deva",
    "guj_Gujr",
    "hin_Deva",
    "kan_Knda",
    "kas_Arab",
    "mai_Deva",
    "mal_Mlym",
    "mar_Deva",
    "mni_Mtei",
    "npi_Deva",
    "ory_Orya",
    "pan_Guru",
    "san_Deva",
    "snd_Deva",
    "tam_Taml",
    "tel_Telu",
    "urd_Arab",
    "khasi",
    "santhali"
]
```

```python
import re

def sentence_tokenize(text):
    text = re.sub(r'\s+', ' ', text).strip()

    sentences = re.split(
        r'(?<=[.!?।॥])\s+',
        text
    )

    sentences = [
        s.strip()
        for s in sentences
        if s.strip()
    ]

    return sentences
```

```python
all_sentences = {}

for lang in languages:

    print(f"\nProcessing: {lang}")

    sentences_for_language = []

    for sample in dataset[lang]:

        text = sample["text"]

        sentences = sentence_tokenize(text)

        for sentence in sentences:

            sentences_for_language.append(sentence)

            # Stop at exactly 1234
            if len(sentences_for_language) == 1234:
                break

        if len(sentences_for_language) == 1234:
            break

    all_sentences[lang] = sentences_for_language

    print(
        f"{lang}: {len(sentences_for_language)} sentences collected"
    )
```

```python
for lang in languages:
    print(f"{lang:12s} : {len(all_sentences[lang])}")
```

```python
import pandas as pd

data = []

for lang in languages:

    for sentence in all_sentences[lang]:

        data.append({
            "text": sentence,
            "label": lang
        })

df = pd.DataFrame(data)

print("Shape:", df.shape)
```

```python
df.to_csv(
    "indic_language_dataset.csv",
    index=False
)
```

```python
import os

size_mb = os.path.getsize(
    "indic_language_dataset.csv"
) / (1024 ** 2)

print(f"Dataset size: {size_mb:.2f} MB")
```

```python
import random

random.seed(42)

train_data = []
val_data = []
test_data = []

for lang in languages:

    sentences = all_sentences[lang].copy()

    # Shuffle sentences within the language
    random.shuffle(sentences)

    # 80 / 10 / 10
    train_sentences = sentences[:800]
    val_sentences = sentences[800:900]
    test_sentences = sentences[900:1000]

    for sentence in train_sentences:
        train_data.append({
            "text": sentence,
            "label": lang
        })

    for sentence in val_sentences:
        val_data.append({
            "text": sentence,
            "label": lang
        })

    for sentence in test_sentences:
        test_data.append({
            "text": sentence,
            "label": lang
        })
```

```python
train_df = pd.DataFrame(train_data)
val_df = pd.DataFrame(val_data)
test_df = pd.DataFrame(test_data)
```

```python
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
print("TRAIN")
print(train_df["label"].value_counts().sort_index())

print("\nVALIDATION")
print(val_df["label"].value_counts().sort_index())

print("\nTEST")
print(test_df["label"].value_counts().sort_index())
```

```python
train_df.to_csv(
    "train.csv",
    index=False
)

val_df.to_csv(
    "validation.csv",
    index=False
)

test_df.to_csv(
    "test.csv",
    index=False
)
```
